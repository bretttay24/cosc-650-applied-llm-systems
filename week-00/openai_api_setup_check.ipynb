### Live Integration Check

This cell loads local environment variables from `.env` and sends a small request to the OpenAI API. A successful response confirms that the local Jupyter environment, dependencies, and API configuration are working. Running it will incur API usage costs.
* **Code Reference:**
    *  [OpenAI, *Migrate to the Responses API*](https://developers.openai.com/api/docs/guides/migrate-to-responses)


```python
# === Imports ===
from dotenv import load_dotenv
from openai import OpenAI

# === Load Environment Variables ===
load_dotenv()

# === Create OpenAI Client ===
client = OpenAI()                           # API key is automatically read from the environment variable OPENAI_API_KEY

# === Generate a Response ===
response = client.responses.create(
    model="gpt-5.6",
    input="Write a one-sentence bedtime story about a unicorn.",
)

# === Print the Response ===
print(response.output_text)
```
